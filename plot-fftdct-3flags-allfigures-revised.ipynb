```python
import numpy as np
import matplotlib.pyplot as plt
import plot_scripts as plts
import os
plt.style.use("mplstyle")
```

```python
def FGname(inp_signal):
    if (inp_signal == '21cm'):
        FG = 'None'
        dataname = '21cm'
    elif (inp_signal == 'FG+21cm'):
        FG = 'Smooth'
        dataname = 'FG'
    elif (inp_signal == 'FG+MM+21cm'):
        FG = 'Unsmooth'
        dataname = 'FGMM'
    return FG, dataname
```

```python
# --------------------------------------------------
path       = "results"
inp_signal = "21cm"
SCF        = "None"

modes  = ['PERIODIC', 'PERIODIC+RANDOM', 'RANDOM']
colors = ['C1', 'C2', 'C3']
fmts = ['s', '^', '>']

FG, dataname = FGname(inp_signal)
shift, ylim = 1.0, None


figname = f"plots/fft_allflags.pdf"
os.makedirs("plots", exist_ok=True)


# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[3, 1, 1, 1],
    hspace=0.07
)

ax1 = fig.add_subplot(gs[0])
axd = [fig.add_subplot(gs[i], sharex=ax1) for i in range(1, 4)]


# --------------------------------------------------
for i, (mode, color, fmt) in enumerate(zip(modes, colors, fmts)):

    # ---------- Load data ----------
    data = np.load(
        f"{path}/fft_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    )

    kb_fft        = data["kb"]
    mean_fft      = data["mean"]
    err_fft       = data["err"]
    binned_th_fft = data["binned_th"]

    inp_signal1 = data["inp_signal"].tolist()
    flag1       = data["flag"].tolist()
    scf1        = data["SCF"].tolist()

    print(f"inp_signal = {inp_signal1}, flag = {flag1}, scf = {scf1}")

    # ---------- TOP PANEL: power spectra ----------
    ax1.errorbar(
        kb_fft,
        abs(mean_fft) * shift,
        err_fft * shift,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        ls='--',
        color=color,
        label=fr"$\rm{{ {mode} }}$"
    )

    if i == 0:
        ax1.plot(
            kb_fft,
            binned_th_fft * shift,
            '-k',
            lw=2,
            label=r'${\rm EoR\,Model}$'
        )

    # ---------- Deviation ----------
    ratio = 100.0 * (mean_fft / binned_th_fft - 1.0)
    err_r = 100.0 * err_fft / binned_th_fft

    axd[i].errorbar(
        kb_fft,
        ratio,
        err_r,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        color=color
    )

    # if band:
    axd[i].fill_between(kb_fft, -err_r, err_r,
                        color=color, alpha=0.2,
                        label="_nolegend_")

    axd[i].axhline(0, color='k', ls='--')
    # axd[i].grid(alpha=0.3)
    axd[i].set_ylabel(r'\rm{\%\,deviation}')

    axd[i].text(
        0.02, 0.85,
        fr"$\rm{{ {mode} }}$",
        transform=axd[i].transAxes,
    )

    if ylim is not None:
        axd[i].set_ylim(-ylim, ylim)


# --------------------------------------------------
ax1.set_yscale('log')
ax1.set_ylabel(r'$\mid P(k_\parallel) \mid \,{\rm mK}^2\,{\rm Mpc}$')
# ax1.grid(alpha=0.3)
ax1.legend()
ax1.tick_params(labelbottom=False)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)

axd[-1].tick_params(labelbottom=True)


axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

ax1.set_title(
    r"\rm{  {\bf FG:}~}" + fr"$\mathrm{{   {FG}   }}$"
    + r"\rm{\quad {\bf Method:}~}" + r'$|\tilde{T}(\tau)|^2$'
    # + r"\rm{\quad {\bf SCF:}~}" + fr"$\mathrm{{{SCF}}}$",
    
)

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
FG, dataname = FGname(inp_signal)
shift, ylim = 1.0, 210

figname = f"plots/{inp_signal}_SCF_{SCF}_allflags.pdf"
os.makedirs("plots", exist_ok=True)

# --------------------------------------------------
fig = plt.figure(figsize=(7, 12))
gs  = fig.add_gridspec(
    6, 1,   # <-- 2 extra panels
    height_ratios=[3, 1, 1, 1, 1.2, 1.2],
    hspace=0.07
)

ax1 = fig.add_subplot(gs[0])
axd = [fig.add_subplot(gs[i], sharex=ax1) for i in range(1, 4)]

# NEW: zoom panels (for last two modes)
axz = [fig.add_subplot(gs[4]), fig.add_subplot(gs[5])]

kmax_zoom = 2

# --------------------------------------------------
for i, (mode, color, fmt) in enumerate(zip(modes, colors, fmts)):

    # ---------- Load data ----------
    data = np.load(
        f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    )

    kb_fft        = data["kb"]
    mean_fft      = data["mean"]
    err_fft       = data["err"]
    binned_th_fft = data["binned_th"]

    print(f"inp_signal = {data['inp_signal'].tolist()}, flag = {data['flag'].tolist()}, scf = {data['SCF'].tolist()}")

    # ---------- TOP PANEL ----------
    ax1.errorbar(
        kb_fft,
        abs(mean_fft) * shift,
        err_fft * shift,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        ls='--',
        color=color,
        ecolor=(*plt.matplotlib.colors.to_rgb(color), 0.3),
        label=fr"$\rm{{ {mode} }}$"
    )

    # mark negative values with 'x'
    negmask = mean_fft < 0
    
    ax1.plot(
        kb_fft[negmask],
        abs(mean_fft[negmask]) * shift,
        marker='x',
        ls='None',
        ms=7,
        mew=1.5,
        # color=color
        color='magenta'
    )

    if i == 0:
        ax1.plot(
            kb_fft,
            binned_th_fft * shift,
            '-k',
            lw=2,
            label=r'${\rm EoR\,Model}$'
        )

    # ---------- Deviation ----------
    ratio = 100.0 * (mean_fft / binned_th_fft - 1.0)
    err_r = 100.0 * err_fft / binned_th_fft

    axd[i].errorbar(
        kb_fft, ratio, err_r,
        ms=4, elinewidth=1,
        fmt=fmt, color=color
    )

    axd[i].fill_between(
        kb_fft, -err_r, err_r,
        color=color, alpha=0.2
    )

    axd[i].axhline(0, color='k', ls='--')
    axd[i].set_ylabel(r'\rm{\%\,deviation}')

    axd[i].text(
        0.02, 0.85,
        fr"$\rm{{ {mode} }}$",
        transform=axd[i].transAxes,
    )

    if ylim is not None:
        axd[i].set_ylim(-ylim, ylim)

    axd[0].set_ylim(-52, 52)

    # ---------- ZOOM PANELS (last two modes only) ----------
    if i in [1, 2]:
        j = i - 1  # map to axz[0], axz[1]

        mask = kb_fft <= kmax_zoom

        axz[j].errorbar(
            kb_fft[mask],
            ratio[mask],
            err_r[mask],
            ms=4,
            elinewidth=1,
            fmt=fmt,
            color=color
        )

        axz[j].fill_between(
            kb_fft[mask],
            -err_r[mask],
            err_r[mask],
            color=color,
            alpha=0.2
        )

        axz[j].axhline(0, color='k', ls='--')
        axz[j].set_xlim(0, 2)
        axz[j].set_ylim(-30, 30)

        axz[j].set_ylabel(r'\rm{\%\,deviation}')
        axz[j].text(
            0.02, 0.85,
            fr"$\rm{{ {mode} }}$",
            transform=axz[j].transAxes
        )

# --------------------------------------------------
# FORMATTING
# --------------------------------------------------
ax1.set_yscale('log')
ax1.set_ylabel(r'$\mid P(k_\parallel) \mid \,{\rm mK}^2\,{\rm Mpc}$')
ax1.legend()
ax1.tick_params(labelbottom=False)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)

# axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

# zoom panels formatting
axz[0].tick_params(labelbottom=False)
axz[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

# title
ax1.set_title(
    r"\rm{  {\bf FG:}~}" + fr"$\mathrm{{   {FG}   }} $"
    + r"\rm{\quad {\bf Method:}~}" + r'${ \mathcal{F} [ C(\Delta\nu) ] }$'
)

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')

print("Saved:", figname)
```

```python
inp_signal = "FG+21cm"
SCF        = "Hann"

FG, dataname = FGname(inp_signal)
shift, ylim = 1.0, 210

figname = f"plots/{dataname}_{SCF}_allflags.pdf"

# --------------------------------------------------
fig = plt.figure(figsize=(7, 13))
gs  = fig.add_gridspec(
    7, 1,   # <-- +2 panels
    height_ratios=[3, 1, 1, 1, 1, 1.2, 1.2],
    hspace=0.07
)

ax1 = fig.add_subplot(gs[0])
axd = [fig.add_subplot(gs[i], sharex=ax1) for i in range(1, 5)]

# NEW: zoom panels (last two modes)
axz = [fig.add_subplot(gs[5]), fig.add_subplot(gs[6])]

kmax_zoom = 2

# --------------------------------------------------
for i, (mode, color, fmt) in enumerate(zip(modes, colors, fmts)):

    # ---------- Load data ----------
    data = np.load(
        f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    )

    kb_fft        = data["kb"]
    mean_fft      = data["mean"]
    err_fft       = data["err"]
    binned_th_fft = data["binned_th"]

    inp_signal1 = data["inp_signal"].tolist()
    flag1       = data["flag"].tolist()
    scf1        = data["SCF"].tolist()

    print(f"inp_signal = {inp_signal1}, flag = {flag1}, scf = {scf1}")

    # ---------- TOP PANEL ----------
    ax1.errorbar(
        kb_fft,
        abs(mean_fft) * shift,
        err_fft * shift,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        ls='--',
        color=color,
        ecolor=(*plt.matplotlib.colors.to_rgb(color), 0.3),
        label=fr"$\rm{{ {mode} }}$"
    )

    # mark negative values with 'x'
    negmask = mean_fft < 0
    
    ax1.plot(
        kb_fft[negmask],
        abs(mean_fft[negmask]) * shift,
        marker='x',
        ls='None',
        ms=7,
        mew=1.5,
        # color=color
        color='magenta'
    )


    if i == 0:
        ax1.plot(
            kb_fft,
            binned_th_fft * shift,
            '-k',
            lw=2,
            label=r'${\rm EoR\,Model}$'
        )

    # ---------- Deviation ----------
    ratio = 100.0 * (mean_fft / binned_th_fft - 1.0)
    err_r = 100.0 * err_fft / binned_th_fft

    axd[i].errorbar(
        kb_fft,
        ratio,
        err_r,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        color=color
    )
    
    axd[i].fill_between(
        kb_fft, -err_r, err_r,
        color=color, alpha=0.2,
        label="_nolegend_"
    )

    axd[i].axhline(0, color='k', ls='--')
    axd[i].set_ylabel(r'\rm{\%\,deviation}')

    axd[i].text(
        0.02, 0.85,
        fr"$\rm{{ {mode} }}$",
        transform=axd[i].transAxes,
    )

    if ylim is not None:
        axd[i].set_ylim(-ylim, ylim)

    axd[0].set_ylim(-12, 12)
    axd[1].set_ylim(-52, 52)

    # ---------- ZOOM PANELS (last two modes only) ----------
    if i in [2, 3]:
        j = i - 2  # map → axz[0], axz[1]

        mask = kb_fft <= kmax_zoom

        axz[j].errorbar(
            kb_fft[mask],
            ratio[mask],
            err_r[mask],
            ms=4,
            elinewidth=1,
            fmt=fmt,
            color=color
        )

        axz[j].fill_between(
            kb_fft[mask],
            -err_r[mask],
            err_r[mask],
            color=color,
            alpha=0.2
        )

        axz[j].axhline(0, color='k', ls='--')
        axz[j].set_xlim(0, 2)
        axz[j].set_ylim(-30, 30)

        axz[j].set_ylabel(r'\rm{\%\,deviation}')
        axz[j].text(
            0.02, 0.85,
            fr"$\rm{{ {mode} }}$",
            transform=axz[j].transAxes
        )

# --------------------------------------------------
# FORMATTING
# --------------------------------------------------
ax1.set_yscale('log')
ax1.set_ylabel(r'$\mid P(k_\parallel) \mid \,{\rm mK}^2\,{\rm Mpc}$')
ax1.legend()
ax1.tick_params(labelbottom=False)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)

axd[-1].tick_params(labelbottom=True)
# axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

# zoom panels formatting
axz[0].tick_params(labelbottom=False)
axz[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

ax1.set_title(
    r"\rm{{\bf FG:}~}" + fr"$\mathrm{{{FG}}}$"
    + r"\rm{\quad {\bf Method:}~}" + r'${ \mathcal{F} [ C(\Delta\nu) ] }$'
    + r"\rm{\quad {\bf SCF:}~}" + fr"$\mathrm{{{SCF}}}$",
)

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')

print("Saved:", figname)
```
