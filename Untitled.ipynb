```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(0)

# Frequency axis
nu = np.linspace(50, 100, 300)

# Shared 21-cm signal (absorption-like feature)
T21 = -0.15 * np.exp(-(nu - 75)**2 / (2 * 4**2))

# Smooth foreground generator
def foreground(scale, phase):
    return scale * (nu / 75)**(-2.5) + 0.02 * np.sin(0.15 * nu + phase)

# Generate multiple spectra
Ns = 4
spectra = []
foregrounds = []

for i in range(Ns):
    fg = foreground(scale=1.0 + 0.1*i, phase=i)
    noise = 0.01 * np.random.randn(len(nu))
    spectra.append(fg + T21 + noise)
    foregrounds.append(fg)

# --- Plot ---
fig = plt.figure(figsize=(10, 8))

# --------------------
# Top row: Observed spectra
# --------------------
for i in range(Ns):
    ax = plt.subplot(3, Ns, i+1)
    ax.plot(nu, spectra[i], color="k", lw=1)
    ax.plot(nu, foregrounds[i], color="orange", lw=2)
    ax.plot(nu, T21, color="navy", lw=2)
    ax.set_title(f"Spectrum {i+1}", fontsize=10)
    ax.set_xticks([])
    ax.set_yticks([])
    if i == 0:
        ax.set_ylabel("Temperature")

# --------------------
# Middle row: Hierarchy explanation
# --------------------
ax = plt.subplot(3, 1, 2)
for i in range(Ns):
    ax.plot(nu, foregrounds[i], color="orange", alpha=0.5)
ax.plot(nu, T21, color="navy", lw=3, label="Shared $T_{21}$")
ax.set_title("Hierarchical model: independent foregrounds + shared 21-cm signal")
ax.set_xticks([])
ax.set_yticks([])
ax.legend(frameon=False)

# --------------------
# Bottom row: Foregrounds removed
# --------------------
ax = plt.subplot(3, 1, 3)
residual = spectra[0] - foregrounds[0]
ax.plot(nu, residual, color="navy", lw=2)
ax.axhline(0, color="gray", ls="--", lw=1)
ax.set_title("After foreground removal: cosmological signal preserved")
ax.set_xlabel("Frequency [MHz]")
ax.set_ylabel("Residual")

plt.tight_layout()
plt.show()
```
