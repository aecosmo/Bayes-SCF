```python
import numpy as np
import matplotlib.pyplot as plt
import plot_scripts as plts
import os
plt.style.use("mplstyle")
```

```python
def FGname(inp_signal):
    if (inp_signal == '21cm'):
        FG = 'None'
        dataname = '21cm'
    elif (inp_signal == 'FG+21cm'):
        FG = 'Smooth'
        dataname = 'FG'
    elif (inp_signal == 'FG+MM+21cm'):
        FG = 'Unsmooth'
        dataname = 'FGMM'
    return FG, dataname
```

```python
# --------------------------------------------------
path       = "results"
inp_signal = "21cm"
SCF        = "None"

modes  = ['PERIODIC', 'PERIODIC+RANDOM', 'RANDOM']
colors = ['C1', 'C2', 'C3']
fmts = ['s', '^', '>']

FG, dataname = FGname(inp_signal)
shift, ylim = 1.0, None


figname = f"plots/fft_allflags.pdf"
os.makedirs("plots", exist_ok=True)


# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[3, 1, 1, 1],
    hspace=0.07
)

ax1 = fig.add_subplot(gs[0])
axd = [fig.add_subplot(gs[i], sharex=ax1) for i in range(1, 4)]


# --------------------------------------------------
for i, (mode, color, fmt) in enumerate(zip(modes, colors, fmts)):

    # ---------- Load data ----------
    data = np.load(
        f"{path}/fft_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    )

    kb_fft        = data["kb"]
    mean_fft      = data["mean"]
    err_fft       = data["err"]
    binned_th_fft = data["binned_th"]

    inp_signal1 = data["inp_signal"].tolist()
    flag1       = data["flag"].tolist()
    scf1        = data["SCF"].tolist()

    print(f"inp_signal = {inp_signal1}, flag = {flag1}, scf = {scf1}")

    # ---------- TOP PANEL: power spectra ----------
    ax1.errorbar(
        kb_fft,
        abs(mean_fft) * shift,
        err_fft * shift,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        ls='--',
        color=color,
        label=fr"$\rm{{ {mode} }}$"
    )

    if i == 0:
        ax1.plot(
            kb_fft,
            binned_th_fft * shift,
            '-k',
            lw=2,
            label=r'${\rm EoR\,Model}$'
        )

    # ---------- Deviation ----------
    ratio = 100.0 * (mean_fft / binned_th_fft - 1.0)
    err_r = 100.0 * err_fft / binned_th_fft

    axd[i].errorbar(
        kb_fft,
        ratio,
        err_r,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        color=color
    )

    # if band:
    axd[i].fill_between(kb_fft, -err_r, err_r,
                        color=color, alpha=0.2,
                        label="_nolegend_")

    axd[i].axhline(0, color='k', ls='--')
    # axd[i].grid(alpha=0.3)
    axd[i].set_ylabel(r'\rm{\%\,deviation}')

    axd[i].text(
        0.02, 0.85,
        fr"$\rm{{ {mode} }}$",
        transform=axd[i].transAxes,
    )

    if ylim is not None:
        axd[i].set_ylim(-ylim, ylim)


# --------------------------------------------------
ax1.set_yscale('log')
ax1.set_ylabel(r'$\mid P(k_\parallel) \mid \,{\rm mK}^2\,{\rm Mpc}$')
# ax1.grid(alpha=0.3)
ax1.legend()
ax1.tick_params(labelbottom=False)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)

axd[-1].tick_params(labelbottom=True)


axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

ax1.set_title(
    r"\rm{  {\bf FG:}~}" + fr"$\mathrm{{   {FG}   }}$"
    + r"\rm{\quad {\bf Method:}~}" + r'$|\tilde{T}(\tau)|^2$'
    # + r"\rm{\quad {\bf SCF:}~}" + fr"$\mathrm{{{SCF}}}$",
    
)

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
FG, dataname = FGname(inp_signal)
shift, ylim = 1.0, 210

figname = f"plots/{inp_signal}_SCF_{SCF}_allflags.pdf"
os.makedirs("plots", exist_ok=True)

# --------------------------------------------------
# FIGURE LAYOUT (full range only, no zoom panels)
# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[3, 1, 1, 1],
    hspace=0.07
)

ax1 = fig.add_subplot(gs[0])
axd = [fig.add_subplot(gs[i], sharex=ax1) for i in range(1, 4)]

# --------------------------------------------------
for i, (mode, color, fmt) in enumerate(zip(modes, colors, fmts)):
    # ---------- Load data ----------
    data = np.load(
        f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    )
    kb_fft        = data["kb"]
    mean_fft      = data["mean"]
    err_fft       = data["err"]
    binned_th_fft = data["binned_th"]
    print(f"inp_signal = {data['inp_signal'].tolist()}, flag = {data['flag'].tolist()}, scf = {data['SCF'].tolist()}")

    # ---------- TOP PANEL ----------
    ax1.errorbar(
        kb_fft,
        (mean_fft) * shift,
        err_fft * shift,
        ms=4,
        elinewidth=1,
        fmt=fmt,
        ls='--',
        color=color,
        ecolor=(*plt.matplotlib.colors.to_rgb(color), 0.3),
        label=fr"$\rm{{ {mode} }}$"
    )

    if i == 0:
        ax1.plot(
            kb_fft,
            binned_th_fft * shift,
            '-k',
            lw=2,
            label=r'${\rm EoR\,Model}$'
        )

    # ---------- Deviation ----------
    ratio = 100.0 * (mean_fft / binned_th_fft - 1.0)
    err_r = 100.0 * err_fft / binned_th_fft

    axd[i].errorbar(
        kb_fft, ratio, err_r,
        ms=4, elinewidth=1,
        fmt=fmt, color=color
    )
    axd[i].fill_between(
        kb_fft, -err_r, err_r,
        color=color, alpha=0.2
    )
    axd[i].axhline(0, color='k', ls='--')
    axd[i].set_ylabel(r'\rm{\%\,deviation}')
    axd[i].text(
        0.02, 0.85,
        fr"$\rm{{ {mode} }}$",
        transform=axd[i].transAxes,
    )

    if ylim is not None:
        axd[i].set_ylim(-ylim, ylim)

    axd[0].set_ylim(-52, 52)

# --------------------------------------------------
# FORMATTING
# --------------------------------------------------
linthresh = 5e2
ax1.set_yscale('log')
ax1.set_yscale('symlog', linthresh=linthresh)
# ax1.set_ylim(-1e3, 1e6)
ax1.axhspan(-linthresh, linthresh, color='grey', alpha=0.15, zorder=0)
ax1.set_ylabel(r'$ P(k_\parallel)  \, {\rm mK}^2\,{\rm Mpc}$')
ax1.legend()
ax1.tick_params(labelbottom=False)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)
axd[-1].tick_params(labelbottom=True)
axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

ax1.set_title(
    r"\rm{  {\bf FG:}~}" + fr"$\mathrm{{   {FG}   }} $"
    + r"\rm{\quad {\bf Method:}~}" + r'${ \mathcal{F} [ C(\Delta\nu) ] }$'
)

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# --------------------------------------------------
# Case A: EoR-only, no SCF
# Case B: EoR-only, Hann-SCF
# Case C: EoR-only, Bayes-SCF
# --------------------------------------------------
path_A = "results"
inp_signal_A = "21cm"
SCF_A = "None"

path_B = "results"
inp_signal_B = "21cm"
SCF_B = "Hann"

path_C = "results"
inp_signal_C = "21cm"
SCF_C = "GP"

modes  = ["NOFLAG", "PERIODIC", "PERIODIC+RANDOM", "RANDOM"]

# distinct color per flagging mode; line style distinguishes case A vs C
fmts   = {"NOFLAG": "o", "PERIODIC": "s", "PERIODIC+RANDOM": "^", "RANDOM": ">"}

figname = "plots/uncertainty_comparison_allflags_overlay.pdf"
os.makedirs("plots", exist_ok=True)

# --------------------------------------------------
# FIGURE LAYOUT: 4 panels, one per flagging mode
# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[1, 1, 1, 1],
    hspace=0.07
)
axd = [fig.add_subplot(gs[0])]
axd += [fig.add_subplot(gs[i], sharex=axd[0]) for i in range(1, 4)]

# Number of lowest k_par bins to skip: these modes are filtered out by
# SCF but not by the no-SCF case, so they are not directly comparable
# between the cases and would bias the comparison.
n_skip = 0

# --------------------------------------------------
def load_err_pct(path, inp_signal, mode, SCF):
    fname = f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    data = np.load(fname)
    kb_fft        = data["kb"][n_skip:]
    mean_fft      = data["mean"][n_skip:]
    err_fft       = data["err"][n_skip:]
    binned_th_fft = data["binned_th"][n_skip:]
    err_r = 100.0 * err_fft # / binned_th_fft
    # err_r = 100.0 * err_fft / mean_fft
    return kb_fft, err_r

# colors distinguish SCF case (no SCF vs Bayes-SCF); markers distinguish flagging mode
color_noscf = 'C0'
color_bayes = 'C1'

# --------------------------------------------------
for i, mode in enumerate(modes):
    ax = axd[i]
    fmt = fmts[mode]

    kA, errA = load_err_pct(path_A, inp_signal_A, mode, SCF_A)
    # kB, errB = load_err_pct(path_B, inp_signal_B, mode, SCF_B)
    kC, errC = load_err_pct(path_C, inp_signal_C, mode, SCF_C)

    ax.plot(kA, np.abs(errA), color=color_noscf, ls='--', marker=fmt, ms=3,
            label="no SCF")
    # ax.plot(kB, np.abs(errB), color=color_B, ls='-.', marker=fmt, ms=3,
    #         label="Hann-SCF")
    ax.plot(kC, np.abs(errC), color=color_bayes, ls='-', marker=fmt, ms=3,
            label="Bayes-SCF")

    ax.set_yscale('log')
    ax.set_xscale('log')
    ax.set_ylabel(r'$\mid \%\,\rm{error} \mid$')
    ax.text(
        0.02, 0.88,
        fr"$\rm{{ {mode} }}$",
        transform=ax.transAxes,
        va='top',
    )
    ax.legend( fontsize=8)

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)
axd[-1].tick_params(labelbottom=True)
axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# --------------------------------------------------
# Case A: EoR-only, no SCF
# Case B: EoR-only, Hann-SCF
# Case C: EoR-only, Bayes-SCF
# --------------------------------------------------
path_A = "results"
inp_signal_A = "21cm"
SCF_A = "None"

path_B = "results"
inp_signal_B = "21cm"
SCF_B = "Hann"

path_C = "results"
inp_signal_C = "21cm"
SCF_C = "GP"

modes  = ["NOFLAG", "PERIODIC", "PERIODIC+RANDOM", "RANDOM"]

# distinct marker per flagging mode
fmts   = {"NOFLAG": "o", "PERIODIC": "s", "PERIODIC+RANDOM": "^", "RANDOM": ">"}

figname = "plots/uncertainty_difference_allflags.pdf"
os.makedirs("plots", exist_ok=True)

# --------------------------------------------------
# FIGURE LAYOUT: 4 panels, one per flagging mode
# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[1, 1, 1, 1],
    hspace=0.07
)
axd = [fig.add_subplot(gs[0])]
axd += [fig.add_subplot(gs[i], sharex=axd[0]) for i in range(1, 4)]

# Number of lowest k_par bins to skip: these modes are filtered out by
# SCF but not by the no-SCF case, so they are not directly comparable
# between the cases and would bias the comparison.
n_skip = 0

# --------------------------------------------------
def load_err_pct(path, inp_signal, mode, SCF):
    fname = f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    data = np.load(fname)
    kb_fft        = data["kb"][n_skip:]
    mean_fft      = data["mean"][n_skip:]
    err_fft       = data["err"][n_skip:]
    binned_th_fft = data["binned_th"][n_skip:]
    err_r = 100.0 * err_fft / binned_th_fft
    return kb_fft, err_r

color_diff = 'k'

# --------------------------------------------------
for i, mode in enumerate(modes):
    ax = axd[i]
    fmt = fmts[mode]

    kA, errA = load_err_pct(path_A, inp_signal_A, mode, SCF_A)
    # kB, errB = load_err_pct(path_B, inp_signal_B, mode, SCF_B)
    kC, errC = load_err_pct(path_C, inp_signal_C, mode, SCF_C)

    # Difference: no-SCF minus Bayes-SCF (positive => Bayes-SCF has SMALLER
    # uncertainty than no-SCF at that k_par)
    diff = np.abs(errA) - np.abs(errC)

    ax.axhline(0, color='grey', ls='--', lw=1)
    ax.plot(kA, diff, color=color_diff, ls='-', marker=fmt, ms=3)

    ax.set_xscale('log')
    ax.set_yscale('log')
    ax.set_ylabel(r'$\Delta\,\mid\%\,\rm{error}\mid$')
    ax.text(
        0.02, 0.88,
        fr"$\rm{{ {mode} }}$",
        transform=ax.transAxes,
        va='top',
    )

for ax in axd[:-1]:
    ax.tick_params(labelbottom=False)
axd[-1].tick_params(labelbottom=True)
axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os

# --------------------------------------------------
# Case A: EoR-only, no SCF
# Case C: EoR-only, Bayes-SCF
# --------------------------------------------------
path_A = "results"
inp_signal_A = "21cm"
SCF_A = "None"

path_C = "results"
inp_signal_C = "21cm"
SCF_C = "GP"

modes  = ['NOFLAG', 'PERIODIC', 'PERIODIC+RANDOM', 'RANDOM']

color_A = 'C0'
color_C = 'C1'
label_A = r'$\rm{no\ SCF}$'
label_C = r'$\rm{Bayes-SCF}$'

figname = "plots/binned_th_fft_comparison_allflags.pdf"
os.makedirs("plots", exist_ok=True)

# Number of lowest k_par bins to skip, consistent with the uncertainty
# comparison scripts (these modes are filtered out by SCF).
n_skip = 0

# --------------------------------------------------
def load_binned_th(path, inp_signal, mode, SCF):
    fname = f"{path}/dct_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz"
    data = np.load(fname)
    kb_fft        = data["kb"][n_skip:]
    binned_th_fft = data["binned_th"][n_skip:]
    return kb_fft, binned_th_fft

# --------------------------------------------------
# FIGURE LAYOUT: 4 panels, one per flagging mode
# --------------------------------------------------
fig = plt.figure(figsize=(7, 10))
gs  = fig.add_gridspec(
    4, 1,
    height_ratios=[1, 1, 1, 1],
    hspace=0.07
)
axd = [fig.add_subplot(gs[0])]
axd += [fig.add_subplot(gs[i], sharex=axd[0]) for i in range(1, 4)]

for i, mode in enumerate(modes):
    ax = axd[i]

    kA, thA = load_binned_th(path_A, inp_signal_A, mode, SCF_A)
    kC, thC = load_binned_th(path_C, inp_signal_C, mode, SCF_C)

    # Difference between the two binned_th_fft arrays (should be ~0 if the
    # denominator used in err_r is identical between the no-SCF and
    # Bayes-SCF cases).
    diff = thC - thA
    frac_diff = 100.0 * diff / thA

    ax.plot(kA, frac_diff, color='k', ls='-', marker='o', ms=3)
    ax.axhline(0, color='grey', ls='--', lw=1)

    ax.set_ylabel(r'$\%\,\rm{diff.}$')
    ax.text(
        0.02, 0.88,
        fr"$\rm{{ {mode} }}$",
        transform=ax.transAxes,
        va='top',
    )
    ax.tick_params(labelbottom=False)
axd[-1].tick_params(labelbottom=True)
axd[-1].set_xlabel(r'$k_\parallel\,{\rm Mpc}^{-1}$')

fig.align_ylabels()
fig.savefig(figname, bbox_inches='tight')
print("Saved:", figname)
```

```python
import numpy as np
import matplotlib.pyplot as plt
import plot_scripts as plts
plt.style.use("mplstyle")

dnuc = 40e-3            # MHz
nua = np.load('nua.npy')
N = len(nua)
dnu = np.arange(N) * dnuc


def FGname(inp_signal):
    if (inp_signal == '21cm'):
        FG = 'None'
        dataname = '21cm'
    elif (inp_signal == 'FG+21cm'):
        FG = 'Smooth'
        dataname = 'FG'
    elif (inp_signal == 'FG+MM+21cm'):
        FG = 'Unsmooth'
        dataname = 'FGMM'
    return FG, dataname


cl21 = np.load('cl21.npy')

# --------------------------------------------------
# NOFLAG, EoR-only, before (no SCF) vs after (Bayes-SCF) filtering
# --------------------------------------------------
path = 'results'
inp_signal = '21cm'
mode = 'NOFLAG'
FG, dataname = FGname(inp_signal)

# ---------- before SCF ----------
SCF = 'None'
data = np.load(f"{path}/CL_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz")
cl = data["cl"]
Nrea = cl.shape[0]
cla_noscf = np.mean(cl, axis=0)
cle_noscf = np.std(cl, axis=0) / np.sqrt(Nrea)

# ---------- after Bayes-SCF ----------
SCF = 'GP'
data = np.load(f"{path}/CL_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz")
cl = data["cl"]
cla_bayes = np.mean(cl, axis=0)
cle_bayes = np.std(cl, axis=0) / np.sqrt(Nrea)

# --------------------------------------------------
nc1 = 640

fig, axes = plt.subplots(
    2, 1,
    figsize=(7.5, 7),
    sharex=True,
    gridspec_kw={"height_ratios": [3, 1.5], "hspace": 0.05}
)
ax0, ax1 = axes

# ============================
# Top panel: C(Delta nu), before vs after SCF
# ============================
ax0.errorbar(
    dnu[:nc1], cla_noscf[:nc1], cle_noscf[:nc1],
    color='k', lw=2, ls='-', errorevery=8,
    label=r'$\rm{No\ SCF}$'
)
ax0.errorbar(
    dnu[:nc1], cla_bayes[:nc1], cle_bayes[:nc1],
    color='C1', lw=2, ls='--', errorevery=8,
    label=r'$\rm{Bayes-SCF}$'
)

ax0.set_ylabel(r'$C(\Delta\nu)$')
ax0.legend(frameon=False)
ax0.text(0.02, 0.92, r'$\rm{NOFLAG,\ EoR-only}$', transform=ax0.transAxes)

# ============================
# Bottom panel: Bayes-SCF - no SCF
# ============================
diff = cla_bayes[:nc1] - cla_noscf[:nc1]

ax1.axhline(0, color='gray', ls='--', lw=1)
ax1.plot(dnu[:nc1], diff, color='C1', lw=2)

ax1.set_ylabel(r'$\Delta C$')
ax1.set_xlabel(r'$\Delta\nu\ {\rm [MHz]}$')
ax1.text(0.02, 0.85, r'$\rm{Bayes-SCF - No\ SCF}$', transform=ax1.transAxes)

plt.tight_layout()
plt.savefig("plots/CL_comparison_NOFLAG_before_after_SCF.pdf")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import plot_scripts as plts
plt.style.use("mplstyle")

dnuc = 40e-3            # MHz
nua = np.load('nua.npy')
N = len(nua)
dnu = np.arange(N) * dnuc


def FGname(inp_signal):
    if (inp_signal == '21cm'):
        FG = 'None'
        dataname = '21cm'
    elif (inp_signal == 'FG+21cm'):
        FG = 'Smooth'
        dataname = 'FG'
    elif (inp_signal == 'FG+MM+21cm'):
        FG = 'Unsmooth'
        dataname = 'FGMM'
    return FG, dataname


cl21 = np.load('cl21.npy')

# --------------------------------------------------
# NOFLAG, EoR-only, before (no SCF) vs after (Bayes-SCF) filtering
# --------------------------------------------------
path = 'results'
inp_signal = '21cm'
mode = 'NOFLAG'
FG, dataname = FGname(inp_signal)

# ---------- before SCF ----------
SCF = 'None'
data = np.load(f"{path}/CL_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz")
cl = data["cl"]
Nrea = cl.shape[0]
cla_noscf = np.mean(cl, axis=0)
cle_noscf = np.std(cl, axis=0) / np.sqrt(Nrea)
cle_noscf = cle_noscf / cla_noscf

# ---------- after Bayes-SCF ----------
SCF = 'GP'
data = np.load(f"{path}/CL_inp_signal-{inp_signal}_flag-{mode}_SCF-{SCF}.npz")
cl = data["cl"]
cla_bayes = np.mean(cl, axis=0)
cle_bayes = np.std(cl, axis=0) / np.sqrt(Nrea)
cle_bayes = cle_bayes/cla_bayes

# --------------------------------------------------
nc1 = 640

fig, axes = plt.subplots(
    2, 1,
    figsize=(7.5, 7),
    sharex=True,
    gridspec_kw={"height_ratios": [3, 1.5], "hspace": 0.05}
)
ax0, ax1 = axes

# ============================
# Top panel: uncertainty (std error) on C(Delta nu), before vs after SCF
# ============================
ax0.plot(
    dnu[:nc1], cle_noscf[:nc1],
    color='k', lw=2, ls='-',
    label=r'$\rm{No\ SCF}$'
)
ax0.plot(
    dnu[:nc1], cle_bayes[:nc1],
    color='C1', lw=2, ls='--',
    label=r'$\rm{Bayes-SCF}$'
)

ax0.set_yscale('log')
ax1.set_ylim(-10, 10)
ax0.set_ylabel(r'$\sigma_{C}(\Delta\nu)$')
ax0.legend(frameon=False)
ax0.text(0.02, 0.92, r'$\rm{NOFLAG,\ EoRonly}$', transform=ax0.transAxes)

# ============================
# Bottom panel: ratio, Bayes-SCF / no SCF
# ============================
ratio = cle_bayes[:nc1] / cle_noscf[:nc1]

ax1.axhline(1, color='gray', ls='--', lw=1)
ax1.plot(dnu[:nc1], ratio, color='C1', lw=2)

ax1.set_ylabel(r'$\sigma_{C}^{\rm Bayes}/\sigma_{C}^{\rm no\,SCF}$')
ax1.set_xlabel(r'$\Delta\nu\ {\rm [MHz]}$')

plt.tight_layout()
plt.savefig("plots/CL_uncertainty_comparison_NOFLAG_before_after_SCF.pdf")
plt.show()
```
