# Hierarchical Gaussian Process Demo for Global 21-cm Experiments

This notebook demonstrates the *key idea* behind hierarchical Gaussian Process
foreground filtering for global 21-cm experiments.

The cosmological 21-cm signal is modeled as a **single latent process shared across
multiple spectra**, while foregrounds are modeled as **observation-specific smooth
processes**.

The goal is **foreground removal**, not direct inference of the 21-cm signal.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.linalg import cholesky, solve

np.random.seed(1)
```

We define simple squared-exponential (RBF) kernels to model:

- the shared global 21-cm signal (short correlation length),
- smooth astrophysical foregrounds (long correlation length).


```python
def rbf_kernel(x, ell, sigma2):
    dx = x[:, None] - x[None, :]
    return sigma2 * np.exp(-0.5 * (dx / ell)**2)
```

We simulate multiple sky-averaged spectra observed with different foreground
realizations but sharing the same cosmological 21-cm signal.


```python
nu = np.linspace(50, 100, 250)  # MHz
Nf = len(nu)
Ns = 4  # number of spectra (e.g. beams / LST bins)
```

The global 21-cm signal is modeled as a **single latent Gaussian Process**
shared across all observations.


```python
# ============================================================
# Hierarchical Gaussian Process demo for global 21-cm experiments
# Fully jitter-safe, single-cell version using GPyTorch
# (all autograd-safe for plotting)
# ============================================================

import torch
import gpytorch
import numpy as np
import matplotlib.pyplot as plt

torch.manual_seed(1)
dtype = torch.double

# ------------------------------------------------------------
# Helper: jitter-safe kernel evaluation
# ------------------------------------------------------------
def eval_kernel_with_jitter(kernel, x, jitter_frac=1e-6):
    K = kernel(x).evaluate()
    jitter = jitter_frac * K.diag().mean()
    return K + jitter * torch.eye(K.shape[0], dtype=K.dtype)

# ------------------------------------------------------------
# Frequency axis and setup
# ------------------------------------------------------------
nu = torch.linspace(50, 100, 250, dtype=dtype)  # MHz
Nf = nu.numel()
Ns = 4  # number of spectra

# ------------------------------------------------------------
# Define GP kernels
# ------------------------------------------------------------
kernel_21 = gpytorch.kernels.ScaleKernel(
    gpytorch.kernels.RBFKernel()
)
kernel_21.base_kernel.lengthscale = 3.0
kernel_21.outputscale = 0.02

kernel_fg = gpytorch.kernels.ScaleKernel(
    gpytorch.kernels.RBFKernel()
)
kernel_fg.base_kernel.lengthscale = 30.0
kernel_fg.outputscale = 5.0

# ------------------------------------------------------------
# Draw shared 21-cm signal (hierarchical latent level)
# ------------------------------------------------------------
K21 = eval_kernel_with_jitter(kernel_21, nu)
T21 = torch.distributions.MultivariateNormal(
    torch.zeros(Nf, dtype=dtype), K21
).sample()

# ------------------------------------------------------------
# Draw independent foreground realizations
# ------------------------------------------------------------
Kfg = eval_kernel_with_jitter(kernel_fg, nu)

foregrounds = []
for _ in range(Ns):
    fg = torch.distributions.MultivariateNormal(
        torch.zeros(Nf, dtype=dtype), Kfg
    ).sample()
    foregrounds.append(fg)

# ------------------------------------------------------------
# Generate observed spectra
# ------------------------------------------------------------
noise_std = 0.03
spectra = []

for i in range(Ns):
    noise = noise_std * torch.randn(Nf, dtype=dtype)
    spectra.append(T21 + foregrounds[i] + noise)

spectra = torch.stack(spectra)

# ------------------------------------------------------------
# Build hierarchical covariance matrix
# ------------------------------------------------------------
Ntot = Ns * Nf
K = torch.zeros((Ntot, Ntot), dtype=dtype)

for i in range(Ns):
    for j in range(Ns):
        block = K21 if i != j else (K21 + Kfg)
        K[i*Nf:(i+1)*Nf, j*Nf:(j+1)*Nf] = block

K += noise_std**2 * torch.eye(Ntot, dtype=dtype)
K += 1e-8 * torch.eye(Ntot, dtype=dtype)  # global jitter

# ------------------------------------------------------------
# Hierarchical GP foreground prediction
# ------------------------------------------------------------
y = spectra.reshape(-1)
alpha = torch.linalg.solve(K, y)

K_fg_cross = torch.zeros((Nf, Ntot), dtype=dtype)
K_fg_cross[:, :Nf] = Kfg

fg_pred = K_fg_cross @ alpha
residual = spectra[0] - fg_pred

# ------------------------------------------------------------
# Detach everything for plotting (important)
# ------------------------------------------------------------
nu_np = nu.detach().numpy()
spectra_np = spectra.detach().numpy()
T21_np = T21.detach().numpy()
foregrounds_np = [fg.detach().numpy() for fg in foregrounds]
residual_np = residual.detach().numpy()

# ------------------------------------------------------------
# Plot
# ------------------------------------------------------------
fig, axes = plt.subplots(3, 1, figsize=(8, 8), sharex=True)

# Observed spectra
for i in range(Ns):
    axes[0].plot(nu_np, spectra_np[i], alpha=0.7)
axes[0].set_title("Observed spectra (different foregrounds, shared 21-cm signal)")
axes[0].set_ylabel("Temperature")

# Latent components
axes[1].plot(nu_np, T21_np, color="navy", lw=2, label="Shared $T_{21}$")
for fg in foregrounds_np:
    axes[1].plot(nu_np, fg, color="orange", alpha=0.3)
axes[1].set_title("Hierarchical GP components")
axes[1].legend(frameon=False)

# Residual after foreground removal
axes[2].plot(nu_np, T21_np, color="navy", lw=2, label="Shared $T_{21}$")

axes[2].plot(nu_np, residual_np, color="navy", lw=2)
axes[2].axhline(0, color="gray", ls="--")
axes[2].set_title("After hierarchical foreground removal")
axes[2].set_xlabel("Frequency [MHz]")
axes[2].set_ylabel("Residual")

plt.tight_layout()
plt.show()

# ------------------------------------------------------------
# End of cell
# ------------------------------------------------------------
```
